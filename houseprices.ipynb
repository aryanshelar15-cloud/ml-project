```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
data = pd.read_csv("/content/house_prices_large.csv")
```

```python
X == data.drop("price", axis=1)
Y = data["price"]
model = LinearRegression()
model.fit(X, Y)
pred = model.predict(X)
import numpy as np
mse = mean_squared_error(Y, pred)
rmse = np.sqrt(mse)
print("RMSE:", rmse)
r2 = r2_score(Y, pred)
print("R2 Score:", r2)
```
